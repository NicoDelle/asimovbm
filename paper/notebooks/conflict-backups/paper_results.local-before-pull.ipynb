| Survey block | Policy | Point of view | Description |

### Convert data in numerical and easy to analyze table

````python
# Make a copy so the original table `df` stays unchanged.
df_clean = df.copy()

# Create an empty dictionary where we will store old column names and new column names.
new_column_names = {}

# Go through every column name in the table.
for old_column_name in df_clean.columns:

    # Rename the timestamp column with a short lowercase name.
    if old_column_name == "Timestamp":
        new_column_names[old_column_name] = "timestamp"

    # Rename the long background/familiarity column with a short name.
    elif old_column_name.startswith("Background / Familiarity"):
        new_column_names[old_column_name] = "robotics_familiarity"

    # Rename survey question columns like "Q1.1 ..." into names like "q1_1".
    elif old_column_name.startswith("Q"):
        question_code = old_column_name.split(" ", 1)[0]
        simple_question_name = question_code.lower().replace(".", "_")
        new_column_names[old_column_name] = simple_question_name

# Apply the simpler column names to the copied table.
df_clean = df_clean.rename(columns=new_column_names)
df_clean
````

Now change the values inside the questions into numerical data

````python
# Find all survey question columns. These are now the columns that start with "q".
score_columns = [column for column in df_clean.columns if column.startswith("q")]

# Create the exact scale used in the survey answers.
score_scale = {
    "-3": -3,
    "-2": -2,
    "-1": -1,
    "0": 0,
    "+1": 1,
    "+2": 2,
    "+3": 3,
}

# Convert one answer, for example "+2 Agree (...)" into the number 2.
def convert_answer_to_number(answer):
    answer_text = str(answer)

    for score_text, score_number in score_scale.items():
        if answer_text.startswith(score_text):
            return score_number

    return None


# Go through every survey question column.
for column in score_columns:
    # Apply the conversion function to every answer in this column.
    df_clean[column] = df_clean[column].apply(convert_answer_to_number)

# Show the cleaned table.
df_clean.head()
````

Now remove the timestamp from `df_clean` and make robotics familiarity boolean.

````python
# Remove the timestamp column from df_clean.
df_clean = df_clean.drop(columns=["timestamp"], errors="ignore")

# Convert one familiarity answer into a boolean value.
def familiarity_to_boolean(answer):
    answer_text = str(answer)

    if answer_text == "True":
        return True

    if answer_text == "False":
        return False

    if answer_text.startswith("Daily interaction"):
        return True

    if answer_text.startswith("No daily contact"):
        return False

    return None


# Apply the boolean conversion to the robotics familiarity column.
df_clean["robotics_familiarity"] = (
    df_clean["robotics_familiarity"]
    .apply(familiarity_to_boolean)
    .astype("boolean")
)

# Show the cleaned table.
df_clean.head()
````

## Graphs

````python
# Import matplotlib for plotting graphs.
from pathlib import Path

import matplotlib.pyplot as plt

# These are the possible answer scores in the survey.
scale_values = [-3, -2, -1, 0, 1, 2, 3]

# These labels make the graph easier to read.
scale_labels = ["-3", "-2", "-1", "0", "+1", "+2", "+3"]

# Plot the answer distribution for one question, for example "q1_1".
def plot_question(question_name):

    # Check that the question exists in df_clean.
    if question_name not in df_clean.columns:
        print(f"Question not found: {question_name}")
        return

    # Create an empty list where we will store the count for each score.
    answer_counts = []

    # Count how many people selected each value from -3 to +3.
    for score in scale_values:
        count = (df_clean[question_name] == score).sum()
        answer_counts.append(count)

    # Create the bar graph.
    plt.bar(scale_labels, answer_counts)

    # Add a title using the question name.
    plt.title(f"Answers for {question_name}")

    # Add a label under the x-axis.
    plt.xlabel("Answer score")

    # Add a label next to the y-axis.
    plt.ylabel("Number of responses")

    # Display the graph.
    plt.show()


# Example: plot the first question.
plot_question("q1_1")
````

````python
# Plot a pie chart for robotics familiarity.
def plot_robotics_familiarity():

    # Count how many people have daily robotics familiarity.
    familiar_count = (df_clean["robotics_familiarity"] == True).sum()

    # Count how many people do not have daily robotics familiarity.
    not_familiar_count = (df_clean["robotics_familiarity"] == False).sum()

    # Count missing or blank familiarity answers.
    missing_count = df_clean["robotics_familiarity"].isna().sum()

    # Store the pie slice labels.
    labels = ["Daily interaction", "No daily contact", "Missing"]

    # Store the pie slice values.
    values = [familiar_count, not_familiar_count, missing_count]

    # Create the pie chart.
    plt.pie(values, labels=labels, autopct="%1.1f%%")

    # Make the chart circular instead of oval.
    plt.axis("equal")

    # Add a title.
    plt.title("Robotics Familiarity")

    # Display the chart.
    plt.show()


# Example: plot robotics familiarity.
plot_robotics_familiarity()
````

## Statistical Goal of the following section

The goal of this section is not to claim that one policy has an absolute, universal quality score. This follows the logic of exploratory data analysis: before fitting heavier models, we first inspect the structure, spread, outliers, and assumptions in the observed data ([NIST/SEMATECH EDA handbook](https://www.nist.gov/publications/nistsematech-e-handbook-statistical-methods-chapter-1-exploratory-data-analysis); [EPA EDA overview](https://www.epa.gov/caddis/exploratory-data-analysis)). The goal is more modest and more defensible: we want to understand how each policy is **perceived by this group of raters**, and how the distribution of those perceptions changes when we compare policies and points of view.

This is a **relative, distribution-based benchmark**. A policy is not only described by one average value. It is described by the full pattern of ratings it receives: how many people strongly disliked it, how many were neutral, how many liked it, and how consistently it beats another policy when the same participant rated both.

This matters because perception is not a single deterministic quantity. Two people can watch the same robot and honestly disagree. The distribution of ratings is therefore part of the result, not just noise to average away.


### What One Survey Row Means

Each row in the CSV is one participant. Each participant watched all four videos and answered the same four questions for each video.

The four video blocks are:

| Block | Policy | Point of view |
|---|---|---|
| `Q1` | A | First person |
| `Q2` | B | First person |
| `Q3` | A | Third person |
| `Q4` | B | Third person |

Each video is itself a bundle of three episodes:

1. A-to-B movement with no obstacles.
2. A-to-B movement with fixed obstacles.
3. A-to-B movement with moving people.

So the ratings are **not episode-level ratings**. They are ratings of the whole 3-episode video bundle. If a participant gives `Q1.2 = +2`, that means they perceived the safety of the whole `Q1` bundle positively; it does not tell us exactly which of the three episodes caused that opinion.


## What Each Question Measures

For every video block, participants answered four perceptual questions:

| Question suffix | Axis | Interpretation |
|---|---|---|
| `.1` | Competence / control | Did the robot seem capable, stable, and in control? |
| `.2` | Safety | Did the robot feel safe around people or obstacles? |
| `.3` | Awareness | Did the robot seem aware of the surrounding scene? |
| `.4` | Positive impression | Did the participant have a good overall impression? |

These four axes are related, but they are not identical. For example, a robot might seem competent but unsafe, or safe but unimpressive. We therefore keep the axes separate before creating any global summary.

A global score can be useful later, but it must be treated as a summary convenience, not as the raw truth. The raw evidence is the distribution of answers for each axis.


## The Rating Scale Is Ordinal

The survey responses are converted to numbers from `-3` to `+3`:

| Value | Meaning |
|---|---|
| `-3` | Strongly disagree |
| `-2` | Disagree |
| `-1` | Slightly disagree |
| `0` | Neutral |
| `+1` | Slightly agree |
| `+2` | Agree |
| `+3` | Strongly agree |

This is an **ordinal scale**. Ordinal means the values have an order, but the distances between neighboring values are not guaranteed to be psychologically equal. [Sullivan and Artino's discussion of Likert-type scales](https://www.montana.edu/msse/archive/old-files/jgme-5-4-18.pdf) is the source for this caution: Likert responses are ordered categories, and frequency distributions are often more informative than means alone when describing them. We know that `+2` is more positive than `+1`, but we should be careful before assuming that the emotional distance from `+1` to `+2` is exactly the same as from `-3` to `-2`.

Because of that, the safest first analysis is based on:

- counts,
- percentages,
- medians,
- interquartile ranges,
- positive / neutral / negative proportions,
- paired wins, ties, and losses.

Means can still be useful as compact summaries, especially with 70 respondents, but they should not be the only evidence. If a mean is close to zero, it might mean most people were neutral, or it might mean half were very positive and half were very negative. Those are different distributions with different interpretations.


#### Why We Care About Distributions

A distribution tells us how the answers are spread across the possible scores. [NIST describes histograms](https://www.itl.nist.gov/div898/handbook/eda/section3/histogra.htm) as a way to summarize the center, spread, skewness, outliers, and modes of a univariate dataset; [EPA similarly lists histograms, boxplots, cumulative distribution functions, and Q-Q plots](https://www.epa.gov/caddis/exploratory-data-analysis) as basic tools for examining variable distributions. For one condition and one axis, the distribution answers questions like:

- Did most people agree or disagree?
- Was the group polarized?
- Was there a large neutral middle?
- Did only a few people give extreme negative ratings?
- Did the policy improve the whole distribution, or only move a few participants?

For example, two policies could have the same mean score but very different distributions:

| Policy | Distribution shape | Interpretation |
|---|---|---|
| Policy X | Most ratings near `0` | People were mostly neutral. |
| Policy Y | Many ratings at `-3` and many at `+3` | People were polarized. |

If we only report the mean, both policies might look similar. If we report the distribution, we see that they create different user experiences.

This is why the first result should be a table or graph for every `condition × axis`, showing how many participants selected each score from `-3` to `+3`.


## Why This Is A Paired Design

A very important feature of this survey is that every participant rated all four videos. This creates repeated measurements from the same people, so methods for paired or repeated samples are appropriate. [SciPy's Wilcoxon documentation](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.wilcoxon.html) describes the signed-rank test as a non-parametric test for two related paired samples, and [SciPy's Friedman documentation](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.friedmanchisquare.html) describes the Friedman test for repeated samples of the same individuals. This is called a **within-subject** or **paired** design.

That is statistically useful because people use rating scales differently. Some participants are generous and often give high scores. Others are strict and often give low scores. If we compare raw averages only, these personal rating styles can hide the real policy difference.

The paired design lets us ask a better question:

> For the same participant, did Policy A receive a higher rating than Policy B?

So instead of only comparing group means, we compute respondent-level differences:

```text
first-person policy effect = Q1 - Q2
third-person policy effect = Q3 - Q4
Policy A POV effect       = Q1 - Q3
Policy B POV effect       = Q2 - Q4
```

If a participant gives `Q1 = +1` and `Q2 = -2` on the same axis, the paired difference is:

```text
Q1 - Q2 = +1 - (-2) = +3
```

That participant perceived Policy A as better than Policy B for that axis in first-person POV.

This is closer to how people likely evaluated the videos mentally: they saw multiple versions and formed comparisons, even if the survey asked them to rate each video separately.


### What A Paired Difference Distribution Means

For each comparison, we do not get one difference. We get one difference per participant. That gives us a new distribution: the distribution of paired differences.

For example, for the first-person policy comparison:

```text
participant_1: Q1_global - Q2_global
participant_2: Q1_global - Q2_global
...
participant_70: Q1_global - Q2_global
```

This difference distribution tells us how consistently one condition beats another.

Useful summaries are:

| Summary | Meaning |
|---|---|
| Mean difference | Average size and direction of the difference. |
| Median difference | Typical participant-level difference, robust to extremes. |
| Win count | Number of participants where the left condition scored higher. |
| Tie count | Number of participants where both conditions scored the same. |
| Loss count | Number of participants where the left condition scored lower. |
| Win rate | Fraction of participants where the left condition wins. |

The win/tie/loss view is especially natural for ordinal survey data because it does not require us to over-trust exact numeric distances. It asks a simpler question: did this participant rate A above B, equal to B, or below B?


## Relative Perceived Advantage

To turn paired comparisons into one simple relative benchmark score, we can use:

```text
relative perceived advantage = (wins + 0.5 × ties) / total_participants
```

This produces a value between `0` and `1`:

| Value | Interpretation |
|---|---|
| `0.50` | The two conditions are tied overall. |
| `> 0.50` | The left condition is perceived better more often. |
| `< 0.50` | The right condition is perceived better more often. |
| `1.00` | The left condition beats the right condition for everyone. |
| `0.00` | The right condition beats the left condition for everyone. |

The `0.5 × ties` term gives half credit when the participant rated both conditions equally. This is similar in spirit to paired comparison scoring, where a win counts as 1, a tie as 0.5, and a loss as 0.

For this survey, the most important relative comparisons are:

```text
Policy A vs Policy B, first-person POV: Q1 vs Q2
Policy A vs Policy B, third-person POV: Q3 vs Q4
```

This score matches the benchmark idea: a policy's result depends on its position relative to other policies in the same reference set. This is also related to paired-comparison measurement: [Brown and Peterson's USDA report](https://research.fs.usda.gov/treesearch/31863) describes paired comparisons as a method for measuring preference orderings from discrete binary choices and scaling those choices into a preference measure.


## Norm-Referenced Interpretation

The direction we are moving toward is a **norm-referenced benchmark**. The [NCME glossary](https://ncme.org/resources/professional-learning/glossary/) defines norm-referenced score interpretation as interpreting performance by comparison to a specified reference population, and defines percentile rank as the percentage of scores in a specified distribution that fall below a given score. In a norm-referenced system, a score is interpreted relative to a reference distribution. This is how many standardized tests work: the meaning of a score depends on how it compares to the scores of other participants or systems.

For our benchmark, the reference distribution would be the set of policies evaluated under the same survey protocol. A future policy would not only receive an absolute rating; it would be placed relative to the policies already tested.

A simple future version could report something like:

```text
Policy X is at the 75th percentile for perceived safety among tested policies.
```

That means Policy X scored higher than about 75% of the reference policies on perceived safety.

However, our current pilot has only two policies: A and B. That is enough to compare A against B, but it is not enough to build a stable general norm. With only two policies, the reference distribution is too small. So in this notebook, we should describe the result as a **pilot relative comparison**, not as a final norm-referenced benchmark.


## Why We Should Not Fit A Normal Distribution Here

A normal distribution is the bell-shaped distribution used in many statistical models. We are explicitly not assuming normality at this first step; the EDA sources above recommend inspecting the empirical shape before deciding whether distributional assumptions are justified, and [Sullivan and Artino](https://www.montana.edu/msse/archive/old-files/jgme-5-4-18.pdf) caution that Likert-type data is ordinal and should be described with its response distribution. It is useful when data is continuous, symmetric, and generated by many small additive sources of variation.

Our survey data does not naturally have that shape:

- responses can only take seven values: `-3, -2, -1, 0, +1, +2, +3`;
- the scale is ordinal;
- ratings may be skewed;
- ratings may be polarized;
- the endpoints create hard limits.

Because of that, the first analysis should use **empirical distributions**. Empirical means we use the distribution we actually observed, instead of forcing the data into a theoretical shape.

Later, if we collect many policies and need a standardized score, we can consider percentile ranks, z-scores, or latent models. But for the current data, empirical distributions and paired comparisons are simpler, clearer, and more honest.


## What We Can And Cannot Claim From This Survey

### We can claim

- How this participant group rated the four video bundles.
- Whether Policy A was rated higher than Policy B within the same POV.
- Whether first-person and third-person POV changed perception.
- Which perceptual axes show the strongest policy differences.
- Whether ratings are mostly positive, neutral, negative, or polarized.

### We cannot claim yet

- A universal absolute quality score for a policy.
- A stable benchmark percentile for future policies.
- A clean effect of each individual episode, because each video bundles three episodes together.
- A clean effect of robot physicality if the robot body did not vary across videos.
- Full removal of order effects, because this survey showed videos in a fixed order.

The fixed order is especially important. If everyone saw `Q1`, then `Q2`, then `Q3`, then `Q4`, condition effects are partially mixed with order effects. For this pilot, we can still analyze the data, but the paper should mention this limitation. In the next survey, the video order should be randomized or counterbalanced.

The analysis should produce the following outputs.

|---|---|---:|---:|---:|---:|---:|---:|---:|

### 2. Condition summaries

- percent negative,

### 3. Paired policy comparisons

For each axis and for the global score:

```text
Q1 - Q2 = Policy A vs Policy B in first-person POV
Q3 - Q4 = Policy A vs Policy B in third-person POV
```

Report mean difference, median difference, wins, ties, losses, and relative perceived advantage.

### 4. Paired POV comparisons

For each axis and for the global score:

```text
Q1 - Q3 = first-person vs third-person for Policy A
Q2 - Q4 = first-person vs third-person for Policy B
```

This tells us whether POV changes perception.

### 5. Uncertainty

Once the descriptive results are clear, add uncertainty estimates. The simplest approach is bootstrap confidence intervals over participants. [SciPy's bootstrap documentation](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.bootstrap.html) describes the bootstrap as repeatedly resampling the original sample with replacement, recomputing the statistic, and using the resulting bootstrap distribution to form a confidence interval; its `paired=True` option keeps corresponding observations together for paired-sample statistics. The bootstrap repeatedly resamples the 70 participants with replacement and recomputes the statistic. This gives an approximate interval for how stable the result is.

For formal paired tests, use non-parametric repeated-measures tests such as [Wilcoxon signed-rank tests](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.wilcoxon.html) for two-condition paired comparisons, or [Friedman tests](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.friedmanchisquare.html) when comparing all four repeated conditions together. These tests should support the descriptive story, not replace it.


## Interpretation Rule For The Paper

The paper should avoid saying:

> Policy A is objectively good.

A more accurate statement is:

> In this pilot survey, Policy A was perceived more positively than Policy B by the same participants, especially in third-person POV.

And for the benchmark direction:

> These results support a distribution-based benchmark design, where a policy is characterized by its empirical rating distribution and by its paired advantage against other policies in the reference set.

That phrasing keeps the claim aligned with what the data can actually support.

### Local validation metric histograms

These histograms use the same simple count-based reading as the human perception plots above. The scores are normalized from 0 to 1, so the bins show how many computed metric values land in each quality band.

````python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

if "metric_viz_dir" not in globals():
    metric_viz_dir = Path("artifacts/local-validation/metric-visualizations")

condition_summary = pd.read_csv(metric_viz_dir / "condition_summary.csv")

axis_columns = [
    "axis_perceived_dexterity",
    "axis_perceived_safety",
    "axis_perceived_social_awareness",
    "axis_impression",
]

axis_labels = {
    "axis_perceived_dexterity": "dexterity",
    "axis_perceived_safety": "safety",
    "axis_perceived_social_awareness": "social awareness",
    "axis_impression": "impression",
}

score_bins = [0.0, 0.2, 0.4, 0.6, 0.8, 1.000001]
score_bin_labels = ["0.0-0.2", "0.2-0.4", "0.4-0.6", "0.6-0.8", "0.8-1.0"]


def count_scores_by_bin(values):
    numeric_values = pd.to_numeric(values, errors="coerce").dropna().clip(0.0, 1.0)
    binned = pd.cut(
        numeric_values,
        bins=score_bins,
        labels=score_bin_labels,
        include_lowest=True,
        right=False,
    )
    return binned.value_counts().reindex(score_bin_labels, fill_value=0)


def plot_score_histogram(long_table, group_column, score_column, title):
    groups = list(long_table[group_column].dropna().drop_duplicates())
    x_positions = list(range(len(score_bin_labels)))
    bar_width = 0.8 / max(len(groups), 1)

    plt.figure(figsize=(9, 4.5))
    for group_index, group in enumerate(groups):
        group_values = long_table.loc[long_table[group_column] == group, score_column]
        counts = count_scores_by_bin(group_values)
        offsets = [x - 0.4 + (bar_width / 2) + group_index * bar_width for x in x_positions]
        plt.bar(offsets, counts.values, width=bar_width, label=f"{group} (n={len(group_values.dropna())})")

    plt.xticks(x_positions, score_bin_labels)
    plt.xlabel("Normalized metric score")
    plt.ylabel("Number of computed scores")
    plt.title(title)
    plt.legend()
    plt.tight_layout()
    plt.show()


axis_scores_long = condition_summary.melt(
    id_vars=["episode", "policy", "perspective", "terminal_status"],
    value_vars=axis_columns,
    var_name="axis",
    value_name="score",
).dropna(subset=["score"])
axis_scores_long["axis"] = axis_scores_long["axis"].map(axis_labels)

plot_score_histogram(
    axis_scores_long,
    group_column="policy",
    score_column="score",
    title="Computed axis score distribution by policy",
)

plot_score_histogram(
    axis_scores_long,
    group_column="perspective",
    score_column="score",
    title="Computed axis score distribution by perspective",
)

metric_columns = [column for column in condition_summary.columns if column.startswith("metric_")]
metric_scores_long = condition_summary.melt(
    id_vars=["episode", "policy", "perspective", "terminal_status"],
    value_vars=metric_columns,
    var_name="metric",
    value_name="score",
).dropna(subset=["score"])

plot_score_histogram(
    metric_scores_long,
    group_column="policy",
    score_column="score",
    title="Computed feature metric score distribution by policy",
)
````
